```python
import os
from redis import Redis
from dotenv import load_dotenv
```

```python
load_dotenv()
REDIS_URI = os.environ.get("REDIS_URI")

if not REDIS_URI:
    print("REDIS_URI doesnt exist")
    exit()

```

```python
try:
    client = Redis.from_url(REDIS_URI, decode_responses=True)
    
    keys = client.keys("*")
    
    print("Connected to valkey")
    print(f"total keys: {len(keys)}")
    
    if keys:
        print("active slowapi keys:")
        for key in keys:
            value = client.get(key)
            print(f"{key}  ->  (total requests: {value})")
    else:
        print("no data")

except Exception as e:
    print(f"aiven conexion error: {e}")
```
